```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

file_path = '/content/benchmark resutls - stage1.csv'
df = pd.read_csv(file_path)
```

```python
font_path = "/content/Roboto-Regular.ttf"
roboto_font = fm.FontProperties(fname=font_path)
```

```python
df.head()
```

```python
hue_order = ["System 1", "Llama3", "System 2"]

palette = {"System 1": "#ff9603", "Llama3": "lightgray", "System 2": "#5848dc"}

plt.figure(figsize=(6, 6))
sns.boxplot(data=df, x="Algorithm", y="Word Counts", hue="hue", hue_order=hue_order, palette=palette)
plt.legend(loc="upper center",  frameon=False, prop=roboto_font)

plt.xlabel("Algorithm", fontproperties=roboto_font)
plt.ylabel("Word Counts", fontproperties=roboto_font)
plt.title("Stage 1", fontproperties=roboto_font)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm


file_path1 = '/content/benchmark resutls - dpo dif.csv'
df1 = pd.read_csv(file_path1)


file_path2 = '/content/benchmark resutls - simpo dif.csv'
df2 = pd.read_csv(file_path2)


font_path = "/content/Roboto-Regular.ttf"
roboto_font_big = fm.FontProperties(fname=font_path, size=13)
roboto_font_med = fm.FontProperties(fname=font_path, size=13)
roboto_font_small = fm.FontProperties(fname=font_path, size=10)


hue_order = ["System 1", "System 2"]
palette = {"System 1": "#fdda7a", "System 2": "#86d387"}


fig, axes = plt.subplots(1, 2, figsize=(8, 4))


sns.violinplot(data=df1, x="Stage", y="Difference", hue="hue", hue_order=hue_order, palette=palette, ax=axes[0], split=True, inner="quart",     edgecolor="#665c53",)
axes[0].set_title("DPO", fontproperties=roboto_font_big)
axes[0].set_xlabel("", fontproperties=roboto_font_small)
axes[0].set_ylabel("Word Count Difference From Baseline", fontproperties=roboto_font_med)
axes[0].legend(loc="upper center", frameon=False, fontsize=12)
axes[0].tick_params(axis='y', labelsize=12, length=0)
axes[0].tick_params(axis='x', labelsize=14, length=0)



sns.violinplot(data=df2, x="Stage", y="Difference", hue="hue", hue_order=hue_order, palette=palette, ax=axes[1], split=True, inner="quart",     edgecolor="#665c53",)
axes[1].set_title("SIMPO", fontproperties=roboto_font_big)
axes[1].set_xlabel("", fontproperties=roboto_font_small)
axes[1].set_ylabel("", fontproperties=roboto_font_small)
axes[1].get_legend().set_visible(False)
axes[1].tick_params(axis='y', labelsize=12, length=0)
axes[1].tick_params(axis='x', labelsize=14, length=0)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig("/content/combined_plots.png")


plt.show()
```

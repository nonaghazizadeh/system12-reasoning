```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
import os
from PIL import Image

file_path = "/content/benchmark resutls - ratio.csv"
df = pd.read_csv(file_path)

df_cleaned = df.iloc[2:].reset_index(drop=True)
df_cleaned.columns = ['benchmark', 'base', 'dpo_0', 'dpo_25', 'dpo_50', 'dpo_75', 'dpo_100',
                      'simpo_0', 'simpo_25', 'simpo_50', 'simpo_75', 'simpo_100']
df_cleaned = df_cleaned.dropna(subset=['benchmark'])

cols_to_convert = ['base', 'dpo_0', 'dpo_25', 'dpo_50', 'dpo_75', 'dpo_100',
                   'simpo_0', 'simpo_25', 'simpo_50', 'simpo_75', 'simpo_100']
df_cleaned[cols_to_convert] = df_cleaned[cols_to_convert].apply(pd.to_numeric)

x_labels = [
    'system1 100\nsystem2 0',
    'system1 75\nsystem2 25',
    'system1 50\nsystem2 50',
    'system1 25\nsystem2 75',
    'system1 0\nsystem2 100'
]
x_values = [0, 1, 2, 3, 4]

sns.set_theme(style="white")

output_dir = "/content/plots"
os.makedirs(output_dir, exist_ok=True)

image_files = []

for _, row in df_cleaned.iterrows():
    benchmark = row['benchmark']
    base = row['base']
    y1 = [row['dpo_0'], row['dpo_25'], row['dpo_50'], row['dpo_75'], row['dpo_100']]
    y2 = [row['simpo_0'], row['simpo_25'], row['simpo_50'], row['simpo_75'], row['simpo_100']]

    fig, ax = plt.subplots(figsize=(5, 5))
    colors = sns.color_palette("pastel", n_colors=4)

    ax.plot(x_values, y1, marker='o', label='DPO', color=colors[0], linewidth=2.5)
    ax.plot(x_values, y2, marker='o', label='SIMPO', color=colors[1], linewidth=2.5)

    ax.set_xticks(x_values)
    ax.set_xticklabels(x_labels, fontsize=8)
    ax.set_xlim(-0.5, len(x_values) - 0.5)

    ax.set_position([0.125, 0.3, 0.775, 0.6])
    ax.tick_params(axis='y', labelsize=8)

    ax.axhline(y=base, color='black', linestyle='--', linewidth=1)

    ax.set_title(benchmark, fontsize=8, loc='center')

    custom_lines = [
        plt.Line2D([0], [0], color=colors[0], lw=2),
        plt.Line2D([0], [0], color=colors[1], lw=2),
        plt.Line2D([0], [0], color='black', linestyle='--', lw=1)
    ]
    legend_labels = ['DPO', 'SIMPO', 'Llama-3']
    ax.legend(custom_lines, legend_labels, loc='lower right', bbox_to_anchor=(1, 0), frameon=False, fontsize=8)

    output_file = os.path.join(output_dir, f"{benchmark}.png")
    plt.savefig(output_file, dpi=300, bbox_inches='tight')
    plt.close()

    image_files.append(output_file)
    print(f"Saved: {output_file}")

images = [Image.open(img) for img in image_files]

img_width, img_height = images[0].size
columns, rows = 4, 3
combined_width = columns * img_width
combined_height = rows * img_height

combined_image = Image.new('RGB', (combined_width, combined_height), (255, 255, 255))

for idx, img in enumerate(images):
    x_offset = (idx % columns) * img_width
    y_offset = (idx // columns) * img_height
    combined_image.paste(img, (x_offset, y_offset))

final_output_path = os.path.join(output_dir, "combined_plots.png")
combined_image.save(final_output_path, dpi=(300, 300))

print(f"Saved combined image: {final_output_path}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
import os
from PIL import Image
from PIL import Image, ImageDraw, ImageFont
import matplotlib.font_manager as fm
from matplotlib.colors import LinearSegmentedColormap, to_rgba


font_path = "/content/Roboto-Regular.ttf"
roboto_font = fm.FontProperties(fname=font_path)

font_path_bold = "/content/Roboto-Bold.ttf"
roboto_font_bold = fm.FontProperties(fname=font_path_bold, size=18)

font_path_info = "/content/Roboto-Regular.ttf"
roboto_font_info = fm.FontProperties(fname=font_path_info, size=16)

file_path = "/content/benchmark resutls - ratiov2.csv"
df = pd.read_csv(file_path)

df_cleaned = df.iloc[2:].reset_index(drop=True)
df_cleaned.columns = ['benchmark', 'base', 'dpo_0', 'dpo_25', 'dpo_50', 'dpo_75', 'dpo_100',
                      'simpo_0', 'simpo_25', 'simpo_50', 'simpo_75', 'simpo_100']
df_cleaned = df_cleaned.dropna(subset=['benchmark'])

cols_to_convert = ['base', 'dpo_0', 'dpo_25', 'dpo_50', 'dpo_75', 'dpo_100',
                   'simpo_0', 'simpo_25', 'simpo_50', 'simpo_75', 'simpo_100']
df_cleaned[cols_to_convert] = df_cleaned[cols_to_convert].apply(pd.to_numeric)

x_labels = ['100/0', '75/25', '50/50', '25/75', '0/100']
x_values = [0, 1, 2, 3, 4]

sns.set_theme(style="white")

output_dir = "/content/plots"
os.makedirs(output_dir, exist_ok=True)

image_files = []

dpo_color = "#97349b"
simpo_color = "#3e4399"
baseline_color = "#fe9123"

for _, row in df_cleaned.iterrows():
    benchmark = row['benchmark']

    base = row['base']
    y1 = [row['dpo_0'], row['dpo_25'], row['dpo_50'], row['dpo_75'], row['dpo_100']]
    y2 = [row['simpo_0'], row['simpo_25'], row['simpo_50'], row['simpo_75'], row['simpo_100']]

    correlation_dpo = np.corrcoef(x_values, y1)[0, 1]
    correlation_simpo = np.corrcoef(x_values, y2)[0, 1]

    fig, ax = plt.subplots(figsize=(7, 5), constrained_layout=True)
    fig.subplots_adjust(bottom=0.2)
    ax.plot(x_values, y1, marker='o', label='DPO', color=dpo_color, linewidth=2.5)
    ax.plot(x_values, y2, marker='o', label='SIMPO', color=simpo_color, linewidth=2.5)

    ax.set_position([0.125, 0.3, 0.775, 0.6])
    ax.tick_params(axis='y', labelsize=8)

    ax.axhline(y=base, color=baseline_color, linestyle='--', linewidth=2)

    ax.set_title(benchmark, fontsize=25, fontproperties=roboto_font_bold, loc='center')

    custom_lines = [
        plt.Line2D([0], [0], color=dpo_color, lw=2, marker='o'),
        plt.Line2D([0], [0], color=simpo_color, lw=2, marker='o'),
        plt.Line2D([0], [0], color=baseline_color, linestyle='--', lw=2)
    ]
    if benchmark == "MultiArith (Arithmetic)":
        legend_labels = ['DPO', 'SIMPO', 'Llama-3']
        ax.legend(custom_lines, legend_labels, loc='best', frameon=False, prop=roboto_font_info)

    ax.tick_params(axis='y', labelsize=14)
    ax.set_xticks(x_values)
    ax.set_xticklabels(x_labels, fontsize=15)
    if benchmark == "MultiArith (Arithmetic)":
      ax.text(0.05, 0.5 , f'r = {correlation_dpo:.2f}', transform=ax.transAxes,
            fontsize=16, color=dpo_color, fontproperties=roboto_font_bold)
      ax.text(0.05, 0.4, f'r = {correlation_simpo:.2f}', transform=ax.transAxes,
            fontsize=16, color=simpo_color, fontproperties=roboto_font_bold)
    elif benchmark == "Coin (Symbolic)":
      ax.text(0.8, 0.2, f'r = {correlation_dpo:.2f}', transform=ax.transAxes,
            fontsize=16, color=dpo_color, fontproperties=roboto_font_bold)
      ax.text(0.8, 0.1, f'r = {correlation_simpo:.2f}', transform=ax.transAxes,
            fontsize=16, color=simpo_color, fontproperties=roboto_font_bold)
    elif benchmark in ["CSQA (Common Sense)", "Strategy (Common Sense)"]:
      ax.text(0.05, 0.2, f'r = {correlation_dpo:.2f}', transform=ax.transAxes,
            fontsize=16, color=dpo_color, fontproperties=roboto_font_bold)
      ax.text(0.05, 0.1, f'r = {correlation_simpo:.2f}', transform=ax.transAxes,
            fontsize=16, color=simpo_color, fontproperties=roboto_font_bold)
    else:
      ax.text(0.05, 0.9, f'r = {correlation_dpo:.2f}', transform=ax.transAxes,
            fontsize=16, color=dpo_color, fontproperties=roboto_font_bold)
      ax.text(0.05, 0.8, f'r = {correlation_simpo:.2f}', transform=ax.transAxes,
            fontsize=16, color=simpo_color, fontproperties=roboto_font_bold)

    output_file = os.path.join(output_dir, f"{benchmark}.png")
    plt.savefig(output_file, dpi=400, bbox_inches='tight')
    plt.close()

    image_files.append(output_file)
    print(f"Saved: {output_file}")

images = [Image.open(img) for img in image_files]

img_width, img_height = images[0].size
columns, rows = 5, 2
combined_width = columns * img_width
combined_height = rows * img_height

combined_image = Image.new('RGB', (combined_width, combined_height), (255, 255, 255))

for idx, img in enumerate(images):
    x_offset = (idx % columns) * img_width
    y_offset = (idx // columns) * img_height
    combined_image.paste(img, (x_offset, y_offset))

final_output_path = os.path.join(output_dir, "combined_plots.png")
combined_image.save(final_output_path, dpi=(500, 500))

print(f"Final image saved at: {final_output_path}")
```
